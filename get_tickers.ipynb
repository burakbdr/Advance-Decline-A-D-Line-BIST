```python
import requests
import pandas as pd
from io import StringIO
```

```python
url = "https://www.isyatirim.com.tr/tr-tr/analiz/hisse/Sayfalar/Temel-Degerler-Ve-Oranlar.aspx?endeks#page-1"
```

```python
html = requests.get(url)
s=StringIO(html.text)
tickers = list(pd.read_html(s)[2]["Kod"])
```
