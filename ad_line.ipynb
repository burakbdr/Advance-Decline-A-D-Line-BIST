```python
end_date = datetime.now()

start_date = end_date - relativedelta(months=12)


print(f"Today's date: {end_date}")
print(f"Date 3 months ago: {start_date}")
```

```python
def calculate_ad_line(tickers, start_date, end_date):
    """
    Calculate the Advance-Decline Line for given tickers
    
    Parameters:
    -----------
    tickers : list
        List of stock symbols
    start_date : datetime or str
        Start date
    end_date : datetime or str
        End date
        
    Returns:
    --------
    pd.DataFrame
        DataFrame containing the AD Line values
    """
    # Convert datetime objects to string format if needed
    if isinstance(start_date, datetime):
        start_date = start_date.strftime('%Y-%m-%d')
    if isinstance(end_date, datetime):
        end_date = end_date.strftime('%Y-%m-%d')
    
    # Get data for first ticker to establish the date range
    reference_data = yf.download(tickers[0], start=start_date, end=end_date, progress=False)
    date_index = reference_data.index
    
    # Initialize DataFrame with the date index
    changes_df = pd.DataFrame(index=date_index)
    
    print(f"Downloading data for {len(tickers)} stocks...")
    successful_downloads = 0
    
    for ticker in tickers:
        try:
            # Get daily data for each ticker
            stock_data = yf.download(ticker, start=start_date, end=end_date, progress=False)
            
            if not stock_data.empty and len(stock_data) > 1:
                # Calculate daily price change and add to DataFrame
                changes_df[ticker] = stock_data['Close'].pct_change()
                successful_downloads += 1
                
        except Exception as e:
            print(f"Error downloading {ticker}: {e}")
            continue
    
    print(f"Successfully downloaded data for {successful_downloads} stocks")
    
    if changes_df.empty:
        raise ValueError("No data was downloaded successfully. Please check the tickers and date range.")
    
    # Drop the first row which will contain NaN due to pct_change()
    changes_df = changes_df.dropna(how='all')
    
    # Calculate advances and declines for each day
    advances = (changes_df > 0).sum(axis=1)
    declines = (changes_df < 0).sum(axis=1)
    
    # Calculate Net Advances (advances - declines)
    net_advances = advances - declines
    
    # Calculate AD Line (cumulative sum of net advances)
    ad_line = net_advances.cumsum()
    
    # Create result DataFrame
    result = pd.DataFrame({
        'Advances': advances,
        'Declines': declines,
        'Net_Advances': net_advances,
        'AD_Line': ad_line
    })
    
    return result

```

```python
end_date = datetime.now()
start_date = end_date - relativedelta(months=9)

try:
    ad_data = calculate_ad_line(tickers, start_date, end_date)
    
    # Plot the AD Line
    plt.figure(figsize=(12, 6))
    plt.plot(ad_data.index, ad_data['AD_Line'])
    plt.title('BIST Advance-Decline Line')
    plt.xlabel('Date')
    plt.ylabel('AD Line')
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    # Print some statistics
    print("\nAD Line Statistics:")
    print(f"Current AD Line value: {ad_data['AD_Line'].iloc[-1]:.2f}")
    print(f"Highest AD Line value: {ad_data['AD_Line'].max():.2f}")
    print(f"Lowest AD Line value: {ad_data['AD_Line'].min():.2f}")
    
except Exception as e:
    print(f"Error calculating AD Line: {e}")
```

```python
# Download XU100.IS data
xu100 = yf.download('XU100.IS', start=start_date, end=end_date)


# Create figure with two y-axes
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot AD Line
color = 'tab:blue'
ax1.set_xlabel('Date')
ax1.set_ylabel('AD Line' , color=color)
ax1.plot(ad_data.index, ad_data['AD_Line'], color=color, label='AD Line')
ax1.tick_params(axis='y', labelcolor=color)

# Create second y-axis
ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('BIST 100', color=color)
ax2.plot(xu100.index, xu100['Close'], color=color, label='BIST 100')
ax2.tick_params(axis='y', labelcolor=color)

# Add title and adjust layout
plt.title('BIST 100 vs Advance-Decline Line')
fig.tight_layout()

# Add legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.grid(True)
plt.show()
```
